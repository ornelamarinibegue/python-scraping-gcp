```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime

def scrape_pagina12():
    url = "https://www.pagina12.com.ar/"
    response = requests.get(url)
    print(f"Código de estado: {response.status_code}")

    if response.status_code != 200:
        raise Exception(f"Error al acceder al sitio: {response.status_code}")

    soup = BeautifulSoup(response.content, "html.parser")

    # Buscamos los div contenedores (no los h2 directamente)
    articulos = soup.find_all("div", class_="article-title")

    if not articulos:
        print("No se encontraron artículos con la clase 'article-title'.")

    noticias = []
    for articulo in articulos:
        # Extraemos el título del h2 dentro del div
        titulo = articulo.find("h2").get_text(strip=True) if articulo.find("h2") else None

        # Buscamos el <a> dentro del div (si existe)
        link_tag = articulo.find("a")
        link = link_tag["href"] if link_tag else None

        if titulo and link:  # Solo agregamos si hay título y enlace
            noticias.append({
                "titulo": titulo,
                "link": f"https://www.pagina12.com.ar{link}" if not link.startswith("http") else link,
                "fecha": datetime.today().date()
            })
        else:
            print(f"Falta título o enlace en: {titulo}")

    df = pd.DataFrame(noticias)
    return df

if __name__ == "__main__":
    df_noticias = scrape_pagina12()
    print(df_noticias.head(10))

```
